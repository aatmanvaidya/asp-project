# CAMEO — Complete EDA

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from datasets import load_dataset
from collections import Counter
```

```python
%%time
all_datasets = ["cafe", "crema_d", "emns", "emozionalmente", "enterface", "jl_corpus", "mesd",
                "nemo", "oreau", "pavoque", "ravdess", "resd", "subesco"]
ds = load_dataset('amu-cai/CAMEO')
print(ds)
```

```python
for dataset_name in all_datasets:
    dataset = ds[dataset_name]
    print("\n" + "#" * 60)
    print(f'Dataset      : {dataset_name}')
    print(f'Num samples: {len(dataset):,}')
    print(f'Columns    : {dataset.column_names}')
    print("\n")
    print(f'Features   :')
    for col, feat in dataset.features.items():
        print(f'  {col}: {feat}')

    emotions = dataset["emotion"]
    emotion_counts = Counter(emotions)

    print("\n")
    print("Unique emotions:")
    print(list(emotion_counts.keys()))

    print("\nEmotion distribution:")
    for emotion, count in emotion_counts.items():
        print(f"{emotion}: {count}")

    gender_counts = Counter(dataset["gender"])
    print("\nGender distribution:")
    for gender, count in gender_counts.items():
        print(f"{gender}: {count}")

    df = pd.DataFrame({"emotion": dataset["emotion"], "gender": dataset["gender"]})

    fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    sns.countplot(data=df, x='emotion', order=df['emotion'].value_counts().index, palette='muted', ax=axes[0])
    axes[0].set_title(f'{dataset_name} — Emotion Distribution')
    axes[0].tick_params(axis='x', rotation=45)
    axes[0].set_xlabel('Emotion')
    axes[0].set_ylabel('Count')

    sns.countplot(data=df, x='gender', palette='pastel', ax=axes[1])
    axes[1].set_title(f'{dataset_name} — Gender Distribution')
    axes[1].set_xlabel('Gender')
    axes[1].set_ylabel('Count')

    plt.tight_layout()
    plt.show()
```

## Dataset Observations

cafe: Balanced across 7 emotions; neutral at half count. Equal gender split.

crema_d: Largest English dataset. Near-uniform across 6 emotions; neutral slightly underrepresented. Gender balanced.

emns: Single speaker, 8 categories including non-standard 'sarcasm' and 'excitement'. Label harmonisation needed.

emozionalmente: Perfectly balanced, exactly 986 recordings per emotion across 7 categories.

enterface: Balanced across 6 emotions. No neutral category, limiting neutral speech modelling.

jl_corpus: Perfectly balanced across 10 categories, but many non-standard labels (apology, concern, encouragement). Harmonisation required.

mesd: Smallest dataset. Near-perfect balance across 6 standard emotions. Only Spanish dataset in CAMEO.

nemo: Near-balanced across 6 emotions. Slightly more male speakers (2412 vs 2069). Surprise underrepresented.

oreau: Smallest dataset in collection. Perfectly balanced emotions. Heavily male-skewed (377 vs 125 female).

pavoque: Single male speaker only. Neutral heavily overrepresented (3126 vs ~570 others). Non-standard 'poker' label from speech synthesis origins.

ravdess: Most balanced dataset. Perfect gender split (720 each). Neutral at half count by design, no intensity variation for neutral speech.

resd: Gender metadata absent (all ). Moderate emotion imbalance. Non-standard 'enthusiasm' label needs harmonisation.

subesco: Largest and most balanced dataset. Perfect emotion balance (1000 each) and gender balance (3500 each)
