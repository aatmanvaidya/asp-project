## nEMO — Polish Emotional Speech Dataset

```python
ds = load_dataset("amu-cai/CAMEO", split="nemo")
print(ds)
```

```python
df_nemo = pd.DataFrame(ds.remove_columns('audio'))
print("Total recordings:", len(df_nemo))
print("\nEmotions:")
print(df_nemo['emotion'].value_counts())
print("\nGender:")
print(df_nemo['gender'].value_counts())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df_nemo, x='emotion', order=df_nemo['emotion'].value_counts().index, palette='muted', ax=axes[0])
axes[0].set_title('nEMO – Emotion Distribution')
axes[0].set_xlabel('Emotion')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=45)

sns.countplot(data=df_nemo, x='gender', palette='pastel', ax=axes[1])
axes[1].set_title('nEMO – Gender Distribution')
axes[1].set_xlabel('Gender')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

nEMO (Polish, 4481 recordings): The dataset is near-balanced across 6 emotions, with neutral slightly overrepresented and surprise underrepresented. Gender distribution slightly favours male speakers (2412 vs 2069). Despite minor imbalances, nEMO provides reasonable coverage across emotional categories.

## PAVOQUE — German Expressive Speech Dataset

```python
ds_pavoque = load_dataset("amu-cai/CAMEO", split="pavoque")
df_pavoque = pd.DataFrame(ds_pavoque.remove_columns('audio'))
print("Total recordings:", len(df_pavoque))
print("\nEmotions:")
print(df_pavoque['emotion'].value_counts())
print("\nGender:")
print(df_pavoque['gender'].value_counts())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df_pavoque, x='emotion', order=df_pavoque['emotion'].value_counts().index, palette='muted', ax=axes[0])
axes[0].set_title('PAVOQUE – Emotion Distribution')
axes[0].set_xlabel('Emotion')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=45)

sns.countplot(data=df_pavoque, x='gender', palette='pastel', ax=axes[1])
axes[1].set_title('PAVOQUE – Gender Distribution')
axes[1].set_xlabel('Gender')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

PAVOQUE (German, 5442 recordings): The dataset shows significant imbalance. First, all 5442 recordings come from a single male speaker,making it the only single-speaker, single-gender dataset in this collection. This severely limits its generalisability for speaker-independent emotion recognition. Second, the emotion distribution is heavily skewed toward neutral (3126 recordings vs ~570 for all other categories).PAVOQUE includes "poker" as an emotion category, which is not present in any other CAMEO dataset and may complicate cross-dataset label harmonisation.
